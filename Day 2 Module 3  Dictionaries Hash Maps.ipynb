1. Count Frequency of Numbers

```python
numbers = [2, 3, 2, 5, 3, 2, 7]
```

```python
frequency = {}

for x in numbers:
    if x in frequency:
        frequency[x] += 1
    else:
        frequency[x] = 1
print(frequency)
```

```python
#Another method

freq = {}

for x in numbers:
    freq[x] = freq.get(x,0) + 1

print(freq)
```

5. Merge Two Dictionaries

```python
sales_january = {
    "Laptop": 100,
    "Phone": 200
}

sales_february = {
    "Laptop": 150,
    "Tablet": 100
}
```

```python
total_sales = {}

for x,y in sales_january.items():
    total_sales[x] = y

for x,y in sales_february.items():
    total_sales[x] = total_sales.get(x,0) + y
print(total_sales)
```

```python
#another way
total_sales = {}

for x,y in sales_january.items():
    total_sales[x] = y

for x,y in sales_february.items():
    if x in total_sales:
        total_sales[x] += y
    else:
        total_sales[x] = y

print(total_sales)
```

6. Find Duplicate Elements Using a Dictionary

```python
numbers = [10, 20, 30, 20, 40, 10, 50]
```

```python
frequency = {}

for x in numbers:
    if x in frequency:
        frequency[x] += 1
    else:
        frequency[x] = 1

for x,y in frequency.items():
    if y > 1:
        print(x)
        
```

8. Group Values by Key

```python
data = [
    ("India", "Ritesh"),
    ("India", "Amit"),
    ("USA", "John"),
    ("India", "Rahul"),
    ("USA", "Mike")
]
```

```python
grouped = {}

for x,y in data:
    if x not in grouped:
        grouped[x] = []
    grouped[x].append(y)

print(grouped)
```
